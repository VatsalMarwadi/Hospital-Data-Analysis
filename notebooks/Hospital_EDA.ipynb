```python
# Convert Patient ID to Numeric Type
hospital_clean_df["Patient_ID"] = pd.to_numeric(hospital_clean_df["Patient_ID"], errors="coerce")
```

```python
# Fix Invalid Patient ID Values
invalid_id = (hospital_clean_df["Patient_ID"] < 10000) | (hospital_clean_df["Patient_ID"] > 100000)
print("Invalid Patient IDs Found:", invalid_id.sum())
hospital_clean_df.loc[invalid_id, "Patient_ID"] = np.nan
```

```python
# Convert Admit_Date, Discharge_Date, Follow-up Date to Datetime
date_columns = ["Admit_Date", "Discharge_Date", "Followup Date"]
for col in date_columns:
    hospital_clean_df[col] = pd.to_datetime(hospital_clean_df[col], errors="coerce", dayfirst=True)
# hospital_clean_df[date_columns].dtypes
```

```python
# Remove Quotes From Feedback and Convert to Numeric Type
hospital_clean_df["Feedback"] = hospital_clean_df["Feedback"].astype(str).str.replace('"', "")
hospital_clean_df["Feedback"] = pd.to_numeric(hospital_clean_df["Feedback"], errors="coerce")
```

```python
# Clean Billing Amount Format and Convert Type
hospital_clean_df["Billing Amount"] = (
hospital_clean_df["Billing Amount"]
    .astype(str).str.replace(",", "", regex=False)
    .str.replace("₹", "", regex=False)
    .str.replace('"', "", regex=False)
    .str.strip())
hospital_clean_df["Billing Amount"] = pd.to_numeric(hospital_clean_df["Billing Amount"], errors="coerce")
```

```python
# Clean Health Insurance Amount Format and Convert Type
hospital_clean_df["Health Insurance Amount"] = (
hospital_clean_df["Health Insurance Amount"].astype(str)
    .str.replace(",", "", regex=False)
    .str.replace("₹", "", regex=False)
    .str.replace('"', "", regex=False)
    .str.strip())
hospital_clean_df["Health Insurance Amount"] = pd.to_numeric(hospital_clean_df["Health Insurance Amount"], errors="coerce")
```

```python
# Remove Exact Duplicate Records
hospital_clean_df = hospital_clean_df.drop_duplicates()
```

```python
# Handle Placeholder Billing Values
hospital_clean_df["Billing Amount"] = hospital_clean_df["Billing Amount"].replace({9999999: np.nan, 1500000: np.nan, 999999: np.nan})
```

```python
# Handle Invalid Feedback (Must Be Between 0 and 5)
hospital_clean_df.loc[(hospital_clean_df["Feedback"] < 0) | (hospital_clean_df["Feedback"] > 5), "Feedback"] = np.nan
```

```python
# Handle Invalid (Negative) Billing Amounts
hospital_clean_df.loc[hospital_clean_df["Billing Amount"] < 0, "Billing Amount"] = np.nan
```

```python
# Handle Invalid (Negative) Insurance Amounts
hospital_clean_df.loc[hospital_clean_df["Health Insurance Amount"] < 0, "Health Insurance Amount"] = np.nan
```

```python
# Handle Insurance Amount Greater Than Billing Amount
invalid_insurance = hospital_clean_df["Health Insurance Amount"] > hospital_clean_df["Billing Amount"]
hospital_clean_df.loc[invalid_insurance, "Health Insurance Amount"] = np.nan
```

```python
# Fix Discharge Date Earlier Than Admit Date
invalid_discharge = hospital_clean_df["Discharge_Date"] < hospital_clean_df["Admit_Date"]
hospital_clean_df.loc[invalid_discharge, "Discharge_Date"] = pd.NaT
```

```python
# Fix Follow-up Date Earlier Than Discharge Date
invalid_followup = hospital_clean_df["Followup Date"] < hospital_clean_df["Discharge_Date"]
hospital_clean_df.loc[invalid_followup, "Followup Date"] = pd.NaT
```

```python
# Fix Unrealistic Length of Stay (More Than 90 Days)
stay_days_check = (hospital_clean_df["Discharge_Date"] - hospital_clean_df["Admit_Date"]).dt.days
too_long = stay_days_check > 90
print("Unrealistic Stays Found:", too_long.sum())
hospital_clean_df.loc[too_long, "Admit_Date"] = pd.NaT
hospital_clean_df.loc[too_long, "Discharge_Date"] = pd.NaT
```

```python
# Handle Missing Patient ID
hospital_clean_df = hospital_clean_df.dropna(subset=["Patient_ID"])
hospital_clean_df["Patient_ID"] = hospital_clean_df["Patient_ID"].astype(int)
```

```python
# Handle Missing Values in Numeric Columns (Median Imputation)
numeric_columns = ["Feedback", "Billing Amount", "Health Insurance Amount"]
for col in numeric_columns:
    hospital_clean_df[col] = hospital_clean_df[col].fillna(hospital_clean_df[col].median())
```

```python
# Handle Missing Values in Categorical Columns (Mode Imputation)
categorical_columns = ["Diagnosis", "Bed_Occupancy", "Test", "Doctor"]
for col in categorical_columns:
    hospital_clean_df[col] = hospital_clean_df[col].fillna(hospital_clean_df[col].mode()[0])
```

```python
# Handle Missing Admit Date and Discharge Date Together
typical_stay_days = (hospital_clean_df["Discharge_Date"] - hospital_clean_df["Admit_Date"]).dt.days.median()

# Case 1: Admit_Date missing but Discharge_Date known -> work backwards from Discharge_Date
missing_admit = hospital_clean_df["Admit_Date"].isnull() & hospital_clean_df["Discharge_Date"].notnull()
hospital_clean_df.loc[missing_admit, "Admit_Date"] = (
    hospital_clean_df.loc[missing_admit, "Discharge_Date"] - pd.to_timedelta(typical_stay_days, unit="D"))

# Case 2: Both Admit_Date and Discharge_Date missing -> use the overall median Admit_Date
both_missing = hospital_clean_df["Admit_Date"].isnull()
hospital_clean_df.loc[both_missing, "Admit_Date"] = hospital_clean_df["Admit_Date"].median()

# Case 3: Discharge_Date still missing -> now Admit_Date is filled everywhere, so work forwards
missing_discharge = hospital_clean_df["Discharge_Date"].isnull()
hospital_clean_df.loc[missing_discharge, "Discharge_Date"] = (
    hospital_clean_df.loc[missing_discharge, "Admit_Date"] + pd.to_timedelta(typical_stay_days, unit="D"))
```

```python
# Handle Missing Follow-up Date
typical_followup_gap = (hospital_clean_df["Followup Date"] - hospital_clean_df["Discharge_Date"]).dt.days.median()
hospital_clean_df["Followup Date"] = hospital_clean_df["Followup Date"].fillna(
    hospital_clean_df["Discharge_Date"] + pd.to_timedelta(typical_followup_gap, unit="D"))
```

```python
# Re-check Discharge Date vs Admit Date After Filling Missing Dates
invalid_discharge = hospital_clean_df["Discharge_Date"] < hospital_clean_df["Admit_Date"]
hospital_clean_df.loc[invalid_discharge, "Discharge_Date"] = hospital_clean_df["Admit_Date"]
```

```python
# Re-check Follow-up Date vs Discharge Date After Filling Missing Dates
invalid_followup = hospital_clean_df["Followup Date"] < hospital_clean_df["Discharge_Date"]
hospital_clean_df.loc[invalid_followup, "Followup Date"] = hospital_clean_df["Discharge_Date"]
```

```python
# Re-check Insurance vs Billing After Filling Missing Numbers
invalid_insurance = hospital_clean_df["Health Insurance Amount"] > hospital_clean_df["Billing Amount"]
hospital_clean_df.loc[invalid_insurance, "Health Insurance Amount"] = hospital_clean_df["Billing Amount"]
```

```python
# Remove Duplicates Created by Standardization
hospital_clean_df = hospital_clean_df.drop_duplicates()
```
